```python
# Remove Duplicate Records

df = df.drop_duplicates()
```

```python
# Replace N/A and NULL
# replace common invalid values with NaN

df= df.replace(["N/A","NULL"," "],np.nan)
```

```python
# Remove Extra Spaces


df=df.apply(lambda x: x.str.strip() if x.dtype =="str" else x)
```

```python
print(pd.__version__)
```

```python
# standerdise text format
# convert text into proper case

df["Customer_Name"]=df["Customer_Name"].str.title()

df["City"]=df["City"].str.title()

df["State"]=df["State"].str.title()
```

```python
df
```

```python
# validate email addresses
# Keep only valid email records
df=df[df["Email"].str.contains("@",na=False)]
```

```python
# Convert Date column

df["Order_Date"] = pd.to_datetime(
    df["Order_Date"],
    errors="coerce",
    dayfirst=True
)

df["Delivery_Date"] = pd.to_datetime(
    df["Delivery_Date"],
    errors="coerce",
    dayfirst=True
)
```

```python
# Convert Numeric Columns

cols=["Qty","Unit_Price","Discount"]

for c in cols:
    df[c]=pd.to_numeric(df[c],errors="coerce")
```

```python
# Remove invalid Quantity
# Remove records having negative or zero quantity

df=df[df["Qty"]>0]
```

```python
#  Fill Missing Values


#df["Discount"].fillna(0,inplace=True)

df["Discount"]= df["Discount"].fillna(0)
df["Phone"] = df["Phone"].astype("string")
df["Phone"]= df["Phone"].fillna("Unknown")

df["Delivery_Date"]= df["Delivery_Date"].fillna(df["Order_Date"])
```

```python
df
```

```python
# feature engineering
# Total sales

df["Sales"]=df["Qty"]*df["Unit_Price"]
```

```python
# Net Amount
df["Net_Amount"]=df["Sales"]-(df["Sales"]*df["Discount"]/100)
```

```python
# Profit (Example)

df["Profit"]=df["Net_Amount"]*0.20
```

```python
# Order Month

df["Month"]=df["Order_Date"].dt.month_name()
df
```

```python
#Year

df["Year"]=df["Order_Date"].dt.year
```

```python
#Weekday

df["Weekday"]=df["Order_Date"].dt.day_name()
```

```python
# Import Data into Mysql

from sqlalchemy import create_engine, text
from sqlalchemy.engine import URL

connection_url = URL.create(
    "mssql+pyodbc",
    host="localhost",
    database="Ecommerce_project",
    query={
        "driver": "ODBC Driver 18 for SQL Server",
        "Trusted_Connection": "yes",
        "TrustServerCertificate": "yes"
    }
)

engine = create_engine(connection_url)
```

```python
with engine.connect() as conn:
    print(conn.execute(text("SELECT DB_NAME()")).scalar())
```

```python
# Export Dataframe to Mysql

df.to_sql(
    name="orders",
    con=engine,
    schema="dbo",
    if_exists="replace",
    index=False
)
```
